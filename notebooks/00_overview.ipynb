# VeriData Agentic Data Scientist — Phase 1 Multi-Dataset Overview

This notebook summarizes the Phase 1 deterministic pipeline results across all **5 real benchmark datasets** (`insurance`, `telco_churn`, `heart_disease`, `student_performance`, `bank_marketing`) and the **messy synthetic benchmark** (`messy_synthetic`).

Each dataset has its own dedicated notebook in `notebooks/<dataset>_phase1.ipynb` and self-contained report in `outputs/<dataset>/report.md` (and `report.html`).

```python
import json
from pathlib import Path
import pandas as pd
from IPython.display import Markdown, display

ROOT = Path("..").resolve()
OUTPUTS = ROOT / "outputs"
rows = []
for ds_name in ["insurance", "telco_churn", "heart_disease", "student_performance", "bank_marketing", "messy_synthetic"]:
    ds_dir = OUTPUTS / ds_name
    meta = json.loads((ds_dir / "run_meta.json").read_text(encoding="utf-8"))
    prof = json.loads((ds_dir / "profile.json").read_text(encoding="utf-8"))
    ins = json.loads((ds_dir / "insights.json").read_text(encoding="utf-8"))
    rows.append({
        "dataset": ds_name,
        "raw_shape": f"{meta['n_rows_raw']:,} x {meta['n_cols_raw']}",
        "cleaned_conservative": f"{meta['n_rows_cleaned_conservative']:,} x {meta['n_cols_cleaned_conservative']}",
        "cleaned_imputed": f"{meta['n_rows_cleaned_imputed']:,} x {meta['n_cols_cleaned_imputed']}",
        "duplicates_detected": prof["n_duplicate_rows"],
        "raw_missing_pct": f"{prof['missing_matrix_summary']['pct_missing_cells'] * 100:.2f}%",
        "eda_figures": len(list((ds_dir / "figures").glob("*.png"))),
        "auto_insights": len(ins),
        "elapsed_sec": meta["elapsed_seconds"],
    })
summary_df = pd.DataFrame(rows)
display(summary_df)
```
