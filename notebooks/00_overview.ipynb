```python
import json
from pathlib import Path
import pandas as pd
from IPython.display import Markdown, display

ROOT = Path('..').resolve()
OUTPUTS = ROOT / 'outputs'
CLEANED = ROOT / 'data' / 'cleaned'
print('Configured output datasets:', [p.name for p in sorted(OUTPUTS.iterdir()) if p.is_dir()])
```

```python
# Summary table across all processed datasets in outputs/
rows = []
for ds_dir in sorted(OUTPUTS.iterdir()):
    if not ds_dir.is_dir():
        continue
    meta_path = ds_dir / 'run_meta.json'
    prof_path = ds_dir / 'profile.json'
    if meta_path.exists() and prof_path.exists():
        meta = json.loads(meta_path.read_text(encoding='utf-8'))
        prof = json.loads(prof_path.read_text(encoding='utf-8'))
        rows.append({
            'dataset': meta['dataset_name'],
            'raw_rows': meta['n_rows_raw'],
            'raw_cols': meta['n_cols_raw'],
            'cleaned_rows_conservative': meta['n_rows_cleaned_conservative'],
            'cleaned_cols_conservative': meta['n_cols_cleaned_conservative'],
            'cleaned_rows_imputed': meta['n_rows_cleaned_imputed'],
            'duplicate_rows_normalized': prof['n_duplicate_rows'],
            'pct_missing_cells': round(prof['missing_matrix_summary']['pct_missing_cells'] * 100, 2),
            'elapsed_sec': meta['elapsed_seconds'],
        })
display(pd.DataFrame(rows))
```
